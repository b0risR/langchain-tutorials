```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from rich.pretty import pprint as rpprint

load_dotenv()
```

```python
from dataclasses import dataclass


@dataclass
class ColourContext:
    favourite_colour: str = "blue"
    least_favourite_colour: str = "yellow"
    second_colour: str = "gray"
    third_colour: str = "green"
```

```python
agent = create_agent(
    model="gpt-5-nano",
    context_schema=ColourContext
    )
```

```python
messages = HumanMessage(content="What is my favourite colour?")

response = agent.invoke(
    {"messages": [messages]},
    context=ColourContext()
    )
```

```python
rpprint(response, indent_guides=True, expand_all=True)
```

```python
print(response["messages"][-1].content)
```

## Accessing Context

```python
from typing import cast

from langchain.tools import ToolRuntime, tool


@tool
def get_favourite_colour(runtime: ToolRuntime) -> str:
    """Get the favourite colour of the user"""
    context = cast(ColourContext, runtime.context)
    return context.favourite_colour

@tool
def get_least_favourite_colour(runtime: ToolRuntime) -> str:
    """Get the least favourite colour of the user"""
    context = cast(ColourContext, runtime.context)
    return context.least_favourite_colour
```

```python
@tool
def get_second_colour(runtime: ToolRuntime) -> str:
    """Get the second favourite colour of the user"""
    return runtime.context.second_colour

@tool
def get_third_colour(runtime: ToolRuntime[ColourContext]) -> str:
    """Get the third favourite colour of the user"""
    return runtime.context.third_colour
```

```python
agent = create_agent(
    model="gpt-5-nano",
    tools=[get_favourite_colour, get_least_favourite_colour, 
           get_second_colour, get_third_colour],
    context_schema=ColourContext
    )
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColourContext()
    )
```

```python
rpprint(response, indent_guides=True, expand_all=True)
```

```python
print(response["messages"][-1].content)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColourContext(favourite_colour="green")
)
```

```python
rpprint(response, indent_guides=True, expand_all=True)
```

```python
print(response["messages"][-1].content)
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What is my second favourite colour?")]},
    context=ColourContext()
    )

print(response["messages"][-1].content)
```
